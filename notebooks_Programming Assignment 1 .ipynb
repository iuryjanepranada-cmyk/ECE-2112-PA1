## A. Word Rotation Problem

Solution:

```python
def rotate_word(text):
    return text [1:] + text[0]
```

## Test

Example:

```python
rotate_word("python") 
```

```python
rotate_word("logic")
```

```python
rotate_word("Code")
```

```python
rotate_word("A")
```

## B. Username Builder Problem

Solution:

```python
def make_username(first_name, last_name):
    final_first = first_name.lower().replace(" ", "")
    final_last = last_name.lower().replace(" ", "")
    return final_first + "." + final_last
```

## Test

Example:

```python
make_username("Ada", "Lovelace")
```

```python
make_username("Alan", "Turing")
```

```python
make_username("Ana Maria", "De Leon")
```

## C. Bookend Swap Problem 

Solution:

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last] + middle + [first]
```

## Test

Example:

```python
swap_bookends([1, 2, 3, 4, 5, 6])
```

```python
swap_bookends(["red", "green", "blue"])
```

```python
swap_bookends([8,3])
```
